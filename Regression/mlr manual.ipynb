```python
import numpy as np
import matplotlib.pyplot as plt

# House Size
x1 = np.array([1000,1200,1400,1600,1800,2000])

# Bedrooms
x2 = np.array([2,2,3,3,4,4])

# House Price
y = np.array([50,60,72,80,95,110])

# Mean
x1_mean = np.mean(x1)
x2_mean = np.mean(x2)
y_mean = np.mean(y)

# Difference from mean
x1_diff = x1 - x1_mean
x2_diff = x2 - x2_mean
y_diff = y - y_mean

# Coefficients
b1 = np.sum(x1_diff * y_diff) / np.sum(x1_diff ** 2)

b2 = np.sum(x2_diff * y_diff) / np.sum(x2_diff ** 2)

# Intercept
b0 = y_mean - (b1 * x1_mean) - (b2 * x2_mean)

print("Intercept (b0): ", b0)
print("Slope x1 (b1): ", b1)
print("Slope x2 (b2): ", b2)

# Prediction
y_pred = b0 + b1*x1 + b2*x2

# Error
error = y - y_pred

# Metrics
sse = np.sum(error**2)

mse = np.mean(error**2)

rmse = np.sqrt(mse)

r2 = 1 - (sse / np.sum((y - np.mean(y))**2))

print("SSE: ", sse)
print("MSE: ", mse)
print("RMSE: ", rmse)
print("R2 Score: ", r2)

# Plot Graph
plt.scatter(x1, y, label="Actual Price")

plt.plot(x1, y_pred, label="Predicted Price")

plt.xlabel("House Size")
plt.ylabel("House Price")

plt.title("Multiple Linear Regression")

plt.legend()

plt.show()
```

<h1>Using Matrix</h1>

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------------
# X Values
# -----------------------------------

# House Size
x1 = np.array([1000,1200,1400,1600,1800,2000])

# Bedrooms
x2 = np.array([2,2,3,3,4,4])

# -----------------------------------
# Y Values
# -----------------------------------

# House Price
y = np.array([50,60,72,80,95,110])

# Create X Matrix

X = np.array([
    [1,1000,2],
    [1,1200,2],
    [1,1400,3],
    [1,1600,3],
    [1,1800,4],
    [1,2000,4]
])

# Create Y Matrix
Y = np.array([
    [50],
    [60],
    [72],
    [80],
    [95],
    [110]
])

print("X Matrix:\n", X)

print("\nY Matrix:\n", Y)

# -----------------------------------
# X Transpose
# -----------------------------------

XT = X.T

print("\nX Transpose:\n", XT)

# -----------------------------------
# Xᵀ * X
# -----------------------------------

XTX = np.dot(XT, X)

print("\nXᵀ * X:\n", XTX)

# Determinant of XᵀX
det = np.linalg.det(XTX)

print("\nDeterminant:\n", det)

# Inverse of XᵀX
XTX_inv = np.linalg.inv(XTX)

print("\nInverse of XᵀX:\n", XTX_inv)

# Xᵀ * Y
XTY = np.dot(XT, Y)

print("\nXᵀ * Y:\n", XTY)

# B Matrix

B = np.dot(XTX_inv, XTY)

print("\nB Matrix:\n", B)

# Coefficients
b0 = B[0][0]
b1 = B[1][0]
b2 = B[2][0]

print("\nIntercept (b0): ", b0)

print("Slope x1 (b1): ", b1)

print("Slope x2 (b2): ", b2)

# Prediction
y_pred = b0 + b1*x1 + b2*x2

# Error
error = y - y_pred

# Metrics
sse = np.sum(error**2)

mse = np.mean(error**2)

rmse = np.sqrt(mse)

r2 = 1 - (sse / np.sum((y - np.mean(y))**2))

print("\nSSE: ", sse)

print("MSE: ", mse)

print("RMSE: ", rmse)

print("R2 Score: ", r2)

# Plot Graph
plt.scatter(x1, y, label="Actual Price")

plt.plot(x1, y_pred, label="Predicted Price")

plt.xlabel("House Size")

plt.ylabel("House Price")

plt.title("Multiple Linear Regression")

plt.legend()

plt.show()
```
