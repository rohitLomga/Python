```python
import numpy as np
import matplotlib.pyplot as plt

# Dataset
# Experience
x = np.array([1, 2, 3, 4, 5], dtype=float)

# Salary in LPA
y = np.array([3, 5, 7, 9, 11], dtype=float)

# Parameters
b0 = 0
b1 = 0

learning_rate = 0.01
epochs = 1000

n = len(x)

# Training
for i in range(epochs):

    y_pred = b0 + b1*x

    error = y - y_pred

    db0 = (-2/n) * np.sum(error)

    db1 = (-2/n) * np.sum(x * error)

    b0 = b0 - learning_rate * db0

    b1 = b1 - learning_rate * db1

# Final Prediction
y_final = b0 + b1*x

# -----------------------------
# Evaluation Metrics
# -----------------------------

# MSE
mse = np.mean((y - y_final) ** 2)

# RMSE
rmse = np.sqrt(mse)

# R² Score
ss_total = np.sum((y - np.mean(y)) ** 2)

ss_residual = np.sum((y - y_final) ** 2)

r2 = 1 - (ss_residual / ss_total)

# Output
print("Intercept:", b0)

print("Slope:", b1)

print("MSE:", mse)

print("RMSE:", rmse)

print("R² Score:", r2)

# Graph
plt.scatter(x, y, label="Actual Data")

plt.plot(x, y_final, label="Regression Line")

plt.xlabel("Experience")

plt.ylabel("Salary")

plt.title("Linear Regression")

plt.legend()

plt.show()
```
