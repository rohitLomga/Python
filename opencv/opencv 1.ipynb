<h3>2. Rotate Image Using Transformation Matrix</h3>

```python
import cv2
import numpy as np

img = cv2.imread('./Photos/tiger-image.jpg')
(h, w) = img.shape[:2]

center = (w//2, h//2)
angle = 45

M = cv2.getRotationMatrix2D(center, angle, 1.0)
rotated = cv2.warpAffine(img, M, (w, h))

cv2.imshow("Rotated", rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

<h3>3. Webcam to Grayscale (Real-Time)</h3>

```python
import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    cv2.imshow("Gray Webcam", gray)

    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```

<h3>5. Convert BGR to HSV</h3>

```python
import cv2

img = cv2.imread('./Photos/tiger-image.jpg')
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

cv2.imshow("Original", img)
cv2.imshow("HSV", hsv)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

<h3>6. Canny Edge Detection (Different Thresholds)</h3>

```python
import cv2

img = cv2.imread('./Photos/tiger-image.jpg', 0)

edges1 = cv2.Canny(img, 50, 150)
edges2 = cv2.Canny(img, 100, 200)

cv2.imshow("Edges 50-150", edges1)
cv2.imshow("Edges 100-200", edges2)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

<h3>7. Count Objects Using Contours</h3>

```python
import cv2

img = cv2.imread('./Photos/tiger-image.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

print("Number of objects:", len(contours))

cv2.drawContours(img, contours, -1, (0,255,0), 2)
cv2.imshow("Contours", img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
