<h3>4 . Write a Python program to print the number of prime numbers that are less than or equal to a given number</h3>

```python
n = int(input("Enter number: "))
count = 0
primes = []

for num in range(2, n + 1):
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        primes.append(num)
        count += 1

for x in primes:
    print(x)

print("Number of prime numbers:", count)
```

<h3>8 . Write a program to identify non prime numbers between 1 and 100. Print the non- prime numbers</h3>

```python
for num in range(1, 101):
    if num < 2:
        print(num, end=" ")
    else:
        for i in range(2, num):
            if num % i == 0:
                print(num, end=" ")
                break
```

<h3>9 . Write a Python program to check if two given numbers are co-prime of not. Return true if two numbers are co-prime otherwise return false</h3>

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

def find_gcd(a, b):
    while b != 0:
        remainder = a % b
        a = b
        b = remainder
    return a

if find_gcd(num1, num2) == 1:
    print("The numbers are co-prime")
else:
    print("The numbers are NOT co-prime")
```

<h3>12 . Write a program to check if a given string contains two similar consecutive letters</h3>

```python
for_similar = input("Enter any string")

similar = False
similar_letter = ""

for i in range(len(for_similar) -1 ):
    if for_similar[i] == for_similar[i + 1]:
        similar = True
        similar_letter = for_similar[i]


print(similar)
if similar:
    print("Similar letter is : ",similar_letter)
else:
    print("No similar letter found")
```
