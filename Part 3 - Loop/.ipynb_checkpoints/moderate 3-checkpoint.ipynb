<h3>2 . Write a program that accepts a word from the user and reverse it.</h3>

```python
word = input("Enter any string")

reversed_word = word[::-1]

print("Reversed String is ",reversed_word)
```

<h3>3 . Write a python program to create the multiplication table (from 1 to 10) of a number.</h3>

```python
for i in range(1,11):
    for x in range(1,11):
        product = i * x
        print(f"{i} X {x} = {product}")
    print()
```

<h3>4 . Write a program to find those numbers which are divisible by 7 and multiples of 5, between 1500 and 2700 (both included).</h3>

```python
for x in range(1500,27001):
    if x % 7 == 0 and x % 5 == 0:
        print(x)
```

<h3>5 . Write a python program to count the number of even and odd numbers in a series of numbers.</h3>

```python
count_even = 0
count_odd = 0

numbers = [5, 4, 8, 6, 24, 15, 32, 20, 19, 47, 52, 25]

for x in numbers:
    if x % 2 == 0:
        count_even = count_even + 1
    else:
        count_odd += 1

print(f"Total no. of even numbers : {count_even}, and odd numbers : {count_odd}")
```

<h3>6 . Write a python program that prints all the numbers from 0 to 6 except 3 and 6.</h3>

```python
for x in range(7):
    if x == 3 or x == 6: continue
    print(x)
```

<h3>9 . Input a number and calculate the sum of its digits using a while loop</h3>

```python
n = int(input("Enter the number"))
sum_of_no = 0

while n > 0:
    rem = n % 10
    sum_of_no += rem
    n //= 10

print("Sum of all the digits : ",sum_of_no)
```
