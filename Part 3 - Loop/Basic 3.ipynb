<h3>1 . Write a program to verify if a number lies between 10 and 50 using loop</h3>

```python



while True:
    num = int(input("Enter a number: "))
    
    if 10 < num < 50:
        print("The number lies between 10 and 50.")
        break
    else:
        print("The number is NOT between 10 and 50. Try again.")
```

<h3>2 . Write a program that prints numbers from 10 to 1, followed by "Blast off".</h3>

```python
for x in range(10,0,-1):
  print(x)
else:
  print("Blast off")
```

<h3>3 . Write a program to display numbers from 1 to 10 using a for loop</h3>

```python
for x in range(1,11):
    print(x)
```

<h3>4 . Write a program to display even numbers from 1 to 20 using a while loop.</h3>

```python
i = 1

while i <= 20:
    print(i)
    i += 1
```

<h3>5 . Write a program to find the first number divisible by 7 in a range using a break statement.</h3>

```python
initial = 30
end = 50

while initial <= end:
    if(initial % 7 == 0):
        print("Found", initial)
        break

    initial = initial + 1
```

<h3>6 . Write a program to print numbers from 1 to 10, skipping 5 using a continue statement.</h3>

```python
for i in range(1, 11):
    if(i == 5):
        continue

    print(i)
```

<h3>7 . Write a program to print numbers from 1 to 10 and display a message if the loop completes fully using else with a for loop.</h3>

```python
for x in range(1,11):
  print(x)
else:
  print("Ended")
```

<h3>8 . Write a program to display a 3 X 3 grid of numbers using nested loops.</h3>

```python
numbers = [4, 8, 9, 5, 7, 2, 9, 1, 6]

rows = 3
cols = 3
index = 0

for i in range(rows):
    for j in range(cols):
        print(numbers[index], end=" ")
        index += 1
    print()
```
