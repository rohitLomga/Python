<h3>1 . Write a program to check whether a character entered by the user is vowel or a consonant.</h3>

```python
char = 'E'

if char.lower() in 'aeiou':
    print("Vowel")
else:
    print("Consonant")
```

<h3>2 . A company gives a bonus if years of service > 5 and salary > 50,000. Write a program using nested if to check bonus eligibility</h3>

```python
year_of_service = 6
salary = 50_000

if year_of_service > 5:
    if salary > 50_000:
        print("You got Bonus")
    else: print("You are not eligible for the bonus")
else:
    print("You are not eligible for bonus")
```

<h3>3 . Write a program that takes age and gender as input and determines the ticket price: Child (age < 12): Free, Adult Male : 200, Adult Female: 150</h3>

```python
age = int(input("Enter your age"))
gender = input("Enter your gender (Male, Female)")
ticket_price = 0

if age < 12:
    ticket_price = 0
else:
    if gender.lower() == "male":
        ticket_price = 200
    else:
        ticket_price = 150

print(f"Your ticket price is {ticket_price}")
```

<h3>4 . Given three sides of a triangle, write a program using nested if to determine whether it is Equilateral, Isosceles, or Scalene.</h3>

```python
side1, side2, side3 = 14,25,30
triangle_type = ""

if side1 == side2 == side3 : triangle_type = "Equivalent"
elif side1 == side2 or side1 == side3 or side2 == side3: triangle_type = "Isosceles"
else: triangle_type = "Scalene"

print(f"It is {triangle_type} triangle")
```

<h3>5 . Write a program using match/switch to display the day of the week for an input number (1 - 7)</h3>

```python
day = int(input("Enter the day of the week as number"))

match day:
  case 1:
    print("Monday")
  case 2:
    print("Tuesday")
  case 3:
    print("Wednesday")
  case 4:
    print("Thursday")
  case 5:
    print("Friday")
  case 6:
    print("Saturday")
  case 7:
    print("Sunday")
    
```

<h3>6 . Use match/swith to build a simple calculator performing Addition, Subtraction, Multiplication, and Division</h3>

```python
num1 = int(input("Enter the number"))
operator = input("Enter the operator(+,-,/,*) ")
num2 = int(input("Enter the number"))

calculated_value = 0

match operator:
    case '+':
        calculated_value = num1 + num2
    case '-':
        calculated_value = num1 -num2
    case '*':
        calculated_value = num1 * num2
    case '/':
        calculated_value = num1 / num2
    case _:
        print("Enter the correct values")

print(f"Total value is {calculated_value}")
```

<h3>7 . Write a match/switch statement to print the type of the shape for input: circle, square, or triangle, with a default case for invalid input</h3>

```python
shape = input("Enter shape (circle, square, triangle): ").lower()

match shape:
    case "circle":
        print("Shape is a Circle")
    case "square":
        print("Shape is a Square")
    case "triangle":
        print("Shape is a Triangle")
    case _:
        print("Invalid input")
```

<h3>8 . Use match / switch to check a traffic light signal (red, yellow, green) and print the appropriate action</h3>

```python
signal_light = input("Enter the signal of the traffic light (red, yellow, green)").lower()

match signal_light:
    case 'red':
        print("Stop")
    case 'yellow':
        print("Slow down")
    case 'green':
        print("Go")
    case _:
        print("Enter the correct signal light")
```

<h3>9 . Write a program using match/switch and if together to check a menu choice (1 - 3) and perform operations only when the input number is positive.</h3>

```python
# NOT ClEAR ABOUT THIS QUESTION

num = int(input("Enter a positive number (1-3): "))

if num > 0:
    match num:
        case 1:
            print("You selected Option 1")
        case 2:
            print("You selected Option 2")
        case 3:
            print("You selected Option 3")
        case _:
            print("Invalid menu choice")
else:
    print("Number must be positive")
```

<h3>10 . Write a Python program to find the median of three values.</h3>

```python
num1, num2, num3 = 45, 65,10
numbers = [num1, num2, num3]
numbers.sort()

median = numbers[1]

print("Median is:", median)
```

<h3>11 . Write a Python program to get the next day of a given date.</h3>

```python
from datetime import datetime, timedelta

date_input = input("Enter date (DD-MM-YYYY): ")
date_object = datetime.strptime(date_input, "%d-%m-%Y")

next_day = date_object + timedelta(days=1)

print("Next day is:", next_day.strftime("%d-%m-%Y"))
```
