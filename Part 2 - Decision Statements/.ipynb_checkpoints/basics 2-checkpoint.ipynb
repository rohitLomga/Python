<h3>1 . Write a program to check whether a year is a leap year.</h3>

```python
year = 1990
is_leap = False

if year % 4 == 0 and year % 100 != 0: is_leap = True
if year % 100 == 0 and year % 400 == 0: is_leap = True

if is_leap:
    print(f"{year} is Leap Year")
else:
    print(f"{year} is not a leap year")
```

<h3>2 . Write a program to check if a number is even or odd.</h3>

```python
num = 4

if num % 2 == 0: print("Even")
else: print("Odd")
```

<h3>3 . Write a program to find the largest of three numbers.</h3>

```python
num1, num2, num3 = 45, 56, 10
greatest = num1

if num1 > num2 and num1 > num3: greatest = num1
elif num2 > num1 and num2 > num3: greatest = num2
else: greatest = num3

greatest
```

<h3>4 . Write a program to check if a number is divisible by 5 and 11.</h3>

<h3>5 . Wrtie a program to assign grades on marks.</h3>

```python
marks = 60
# grade = 'A'

if marks >= 90: grade = 'A+'
elif marks >= 80 and marks < 90 : grade = 'A'
elif marks >= 70 and marks < 80 : grade = 'B'
elif marks >= 60 and marks < 70 : grade = 'C'
elif marks >= 50 and marks < 60 : grade = 'D'
elif marks < 50 : grade = 'F'

grade
```

<h3>6 . Write a program to classify a number as positive, negative, or zero using if - elif - else</h3>

```python
num = 4

if num > 0: print("Positive")
elif num < 0: print("Negative")
else: print("Zero")
```

<h3>7 . Write a program to check if a number is divisible by both 3 and 5 using nested if-else statements.</h3>

```python
num = 8

if num % 3 == 0:
    if num % 5 == 0:
        print("Yes divisible by both 3 and 5")
    else: print("Not divisible by 5")
else: print("Not divisible by both 3 and 5")
```

<h3>8 . Write a program to determine if a number is even and greater than 10 using logical operators(and, or, not).</h3>

```python
num = 19

if num % 2 == 0 and num > 10 : print("Yes number is even and greater than 10")
elif num % 2 == 0 and num < 10 : print("Number is even but less than 10")
elif num % 2 == 0 and num == 10 : print("Number is even and equals to 10")
else: print("It is ODD number")
```

<h3>9 . Write a program to check if a number is positive using shortcircuiting with logical operators</h3>

```python
num = -10

is_positive = num > 0

is_positive
```

<h3>10 . Write a program to check if a number is even or odd using a ternary operator and display the result.</h3>

```python
num = 9

is_even = True if num % 2 == 0 else False

if is_even : print(f"{num} is even")
else: print(f"{num} is odd")
```
