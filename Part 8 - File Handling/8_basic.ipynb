<h3>1. Open a text file and display its contents.</h3>

```python
with open("details.txt","r") as f:
    print(f.read())
```

<h3>2. Create a new file and write text into it.</h3>

```python
with open("hello.txt","w") as f:
    f.write("Hello File")
```

<h3>3. Read the first line of a file.</h3>

```python
with open("details.txt","r") as f:
    print(f.readline())
```

<h3>4. Read all lines using a loop.</h3>

```python
with open("details.txt","r") as f:
    for line in f:
        print(line.strip())
```

<h3>5. Append data to an existing file.</h3>

```python
with open("hello.txt","a") as f:
    f.write("\nNew line added")
```

<h3>6. Count number of lines in a file.</h3>

```python
with open("details.txt") as f:
    print(len(f.readlines()))
```

<h3>7. Count number of words in a text file.</h3>

```python
with open("details.txt") as f:
    words=f.read().split()
    print(len(words))
```

<h3>8. Copy contents of one file to another.</h3>

```python
with open("details.txt") as f:
    data = f.read()
    with open("hello.txt","w") as f:
        f.write(data)
```

<h3>9. Search for a specific word in a file.</h3>

```python
word = "python"
with open("details.txt") as f:
    text = f.read()
    print(word in text)
```

<h3>10. Display only even numbered lines.</h3>

```python
with open("details.txt") as f:
    lines = f.readlines()
    for i in range(1,len(lines),2):
        print(lines[i].strip())
```

<h3>11. Write user input into a file.</h3>

```python
text = input("Enter text: ")
with open("hello.txt","w") as f:
    f.write(text)
```

<h3>12. Properly open and close file after reading.</h3>

```python
with open("details.txt") as f:
    data = f.read()
    print(data)
```
