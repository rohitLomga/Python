<h3>1. Read a file and display frequency of each word.</h3>

```python
from collections import Counter
with open("details.txt") as f:
    words = f.read().split()
print(Counter(words))
```

<h3>2. Merge contents of two files into a new file.</h3>

```python
with open("details.txt") as f1,open("hello.txt") as f2:
    data = f1.read() + f2.read()
    with open("merged_data.txt","w") as f:
        f.write(data)
```

<h3>3. Copy only lines containing a keyword.</h3>

```python
keyword = "Python"
with open("details.txt") as f, open("merged_data.txt","w") as out:
    for line in f:
        if keyword in line:
            out.write(line)
```

<h3>4. Count vowels and consonants in a file.</h3>

```python
v = c = 0 
with open("details.txt") as f:
    text = f.read().lower()
    for ch in text:
        if ch.isalpha():
            if ch in "aeiou":
                v += 1
            else: 
                c += 1
                
print("Vowels:",v,"Consonants:",c)
```

<h3>5. Remove duplicate lines from a text file.</h3>

```python
with open("details.txt") as f:
    lines = set(f.readlines())
    with open("hello.txt","w") as f:
        f.writelines(lines)
```

<h3>6. Display numbers in right-aligned table.</h3>

```python
nums = list(map(int,input("Enter numbers: ").split()))
for n in nums:
    print(f"{n:10}")
```

<h3>7. Display student report card table.</h3>

```python
students = [("Rohit",85,"A"),("Amit",78,"B")]
print(f"{'Name':10}{'Marks':10}{'Grade':15}")
for s in students:
    print(f"{s[0]:10}{s[1]:1}{s[2]:15}")
```

<h3>8. Convert date DD/MM/YYYY to Month Day, Year.</h3>

```python
import datetime
d = input("Enter date DD/MM/YYYY: ")
dt = datetime.datetime.strptime(d,"%d/%m/%Y")
print(dt.strftime("%B %d, %Y"))
```

<h3>11. Return None if number is negative else square.</h3>

```python
def square(n):
    if n < 0:
        return None
        
    return n*n
```

```python
sq = square(5)
print(sq)
```

<h3>12. Filter None values from list.</h3>

```python
data = [1,None,4,None,5]
print([x for x in data if x is not None])
```

<h3>13. Optional arguments using None default.</h3>

```python
def greet(name=None):
    if name is None:
        print("Hello Guest")
    else:
        print("Hello",name)
```

```python
greet()
greet("Rohit")
```

<h3>14. Replace blank lines with default string.</h3>

```python
with open("details.txt") as f:
    lines = f.readlines()
    with open("hello.txt","w") as f:
        for l in lines:
            f.write(l if l.strip() else "DEFAULT\n")
```

<h3>15. Combine input, file write, read, format table and handle None.</h3>

```python
name = input("Enter name: ")
marks = input("Enter marks: ")
if marks == "":
    marks = None
    with open("students.txt","w") as f:
        f.write(f"{name},{marks}")
        with open("students.txt") as f:
            n, m = f.read().split(",")
            print(f"{'Name':10}{'Marks':10}")
            print(f"{n:10}{m if m!='None' else 'N/A':10}")
```
