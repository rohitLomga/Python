###**Missing value imputation**
In data science and statistics, missing value imputation is the process of replacing missing data (nulls or blanks) with substituted values.

Missing values are data points that are absent or not recorded in a dataset where a value is expected. In other words, for a given variable (column) in your dataset, some entries are empty, null, or undefined.

**Why Missing Values Occur**

Missing values can happen for a variety of reasons, often categorized as technical, human, or contextual factors:

1. Human Error or Oversight

a. People filling out forms may skip questions.

b. Data entry mistakes can result in blank fields.

2. Technical or System Issues

a. Sensors or devices may fail to record readings.

b. Data transfer or storage errors can lead to loss of information.

3. Intentional Non-response

a. Survey participants may choose not to answer sensitive questions (e.g., income, medical history).

4. Inapplicable Data

a. Certain questions may not apply to all cases, e.g., “Number of children” is irrelevant for someone who has no children.

5. Data Integration Issues

a. When combining datasets from multiple sources, some variables might be missing for some records.


###**Types of Missing Data**

**1. Missing Completely at Random (MCAR)**

The missing values happen purely by chance.
There is no relationship between the missing data and any other data (observed or unobserved).

If data is MCAR, you can safely analyze the remaining data without much bias, because the missing part is just random.
The data is missing randomly, like flipping a coin.

Example:

Imagine you are collecting test scores from students, and:

a. Some answer sheets got lost because of a printer error.

b. The loss has nothing to do with how well students performed.

That is MCAR — the missing scores are random and unrelated to performance.



**2 Missing at Random (MAR)**

The missing values are related to other observed variables, but not to the missing value itself.
The missing data depends on some known information.

The data is missing because of some other information we already know,
but not because of the missing value itself.

Example:

Suppose you are studying income and education level.

You notice:

a. People with lower education are less likely to report their income.

b. But among people with the same education level, income missingness is random.

Here:

a. Missing income depends on education (which we know).

b. It does NOT directly depend on income itself.

example 2:

Imagine a hospital survey collecting patients’ income.

Younger patients mostly report their income.

Older patients often skip the income question.

Here:

Income is missing.

Missing depends on age (which we know).

It is not directly because the income is high or low. So this is MAR.

**3 Missing Not at Random (MNAR)**

The missing values depend on the actual missing value itself.
The reason data is missing is directly related to what the missing value would have been.
The missingness depends on the unseen data itself.

People hide information because of what the value actually is.


Example:

Suppose you are collecting salary information.

a. People with very high salaries refuse to report their income.

b. Or people with very low salaries refuse to report it.

Here:

a. Missing income depends on the actual income.

b. The reason is directly related to the missing value.

**MCAR is the easiest to handle.**

**MAR can often be handled with statistical techniques.**

**MNAR is the hardest and can cause serious bias.**

###**Why Care About Missing Values?**

**1. Affect Machine Learning Models**

a. Most machine learning algorithms:

b. Cannot handle missing values directly

c. May give errors

d. May produce poor predictions

So missing data must be treated before training models.

**2. Can Change Conclusions**

Different ways of handling missing data (deleting, filling, estimating) can lead to different results.

So, conclusions may change depending on how missing values are treated.

**3. Reduce Sample Size**

Many statistical methods delete rows with missing values.
This reduces the number of observations.

Smaller sample size:

a. Reduces statistical power

b. Makes results less stable

**4. Reduce Accuracy**

When data is missing, we have less information to work with.
Less information = less accurate results.

Example:
If 30 out of 100 test scores are missing, your class average may not be reliable.

**5. Can Give Wrong Results (Bias)**

If certain types of people or values are missing more than others, the results become unfair or misleading.

Example:
If rich people don’t report income, the average income will look lower than it really is.

**6. May Hide Important Patterns**

Sometimes missing data itself tells a story.

Example:
If many patients skip answering mental health questions, that may indicate discomfort or stigma.

**7. Increase Risk of Misinterpretation**

If researchers ignore missing data, they may:

a. Overestimate effects

b. Underestimate risks

c. Make wrong policy decisions

Only complete rows (with no missing values) are kept for analysis.

Listwise deletion is a method of handling missing data in which all observations with at least one missing value are completely removed from the analysis.

##**2. Simple Imputation — Mean and Mode**

Simple imputation means filling in missing values with a simple replacement value instead of deleting the row.

Two common types:

a. Mean imputation (for numbers) : Replace missing numerical values with the average (mean) of that column.It is used when the variable is numerical (like age, height, salary, marks).

b. Mode imputation (for categories) : Replace missing values with the most frequently occurring value (mode).It is used when the variable is categorical (like gender, color, city).

###**Example : 1**

```python
import pandas as pd
import numpy as np

# Create dataset
data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Marks": [80, 90, np.nan, 70, 60]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)
```

**Apply Mean Imputation**

```python
# Calculate mean
mean_value = df["Marks"].mean()

# Fill missing value with mean
df["Marks"] = df["Marks"].fillna(mean_value)

print("\nAfter Mean Imputation:")
print(df)
```

**Apply Mode Imputation**

```python
# Create dataset
data2 = {
    "Student": ["A", "B", "C", "D", "E"],
    "Gender": ["Male", "Female", "Female", np.nan, "Female"]
}

df2 = pd.DataFrame(data2)

print("\nOriginal Data:")
print(df2)
```

```python
# Calculate mode
mode_value = df2["Gender"].mode()[0]

# Fill missing value with mode
df2["Gender"] = df2["Gender"].fillna(mode_value)

print("\nAfter Mode Imputation:")
print(df2)
```

.mode() finds the most frequent value.

[0] selects the first mode (in case there are multiple).

.fillna(mode_value) replaces missing values.

**3. Linear Interpolation**

Linear interpolation is a method used to fill missing values by drawing a straight line between two known values and estimating the missing value in between.
It assumes the change between two points is steady and straight (linear).
If a value is missing between two numbers,
we fill it by assuming the data moves smoothly in a straight line.

Filling missing values using nearby values.

Example: We know:

Day 1 = 20°C

Day 3 = 24°C

The increase from Day 1 to Day 3 is:

24 − 20 = 4°C

There are 2 days difference, so:

4 ÷ 2 = 2°C increase per day

So:

Day 2 = 20 + 2 = 22°C



###**Example : 1**

```python
import pandas as pd
import numpy as np

# Create dataset with missing value
data = {
    "Day": [1, 2, 3, 4, 5],
    "Temperature": [20, np.nan, 24, np.nan, 28]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)
```

```python
# Perform linear interpolation
df["Temperature"] = df["Temperature"].interpolate(method="linear")

print("\nAfter Linear Interpolation:")
print(df)
```

**4.  Forward/Backward Fill**

These are simple methods to fill missing values using nearby values.

They are mostly used in time-series data (like daily sales, stock prices, temperature).

**Forward fill means:**

Fill the missing value with the previous (above) value.

It “carries forward” the last known value.

“Use the last known value to fill the missing value.”



**Backward fill means:**

 Fill the missing value with the next (below) value.

It “looks forward” and fills backward.

“Use the next known value to fill the missing value.”



*   forward-fill: assume 11:00 and 12:00 were roughly like 10:00,


*   back-fill: assume 11:00 and 12:00 were roughly like 13:00,





###**Example : 1**

```python
import pandas as pd
import numpy as np

# Create dataset with missing values
data = {
    "Day": [1, 2, 3, 4],
    "Sales": [100, np.nan, np.nan, 130]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)
```

**Forward Fill (ffill)**

```python
df_ffill = df.copy()
df_ffill["Sales"] = df_ffill["Sales"].fillna(method="ffill")

print("\nAfter Forward Fill:")
print(df_ffill)
```

**Backward Fill (bfill)**

```python
df_bfill = df.copy()
df_bfill["Sales"] = df_bfill["Sales"].fillna(method="bfill")

print("\nAfter Backward Fill:")
print(df_bfill)
```

**6. KNN Imputation**

KNN imputation is a smart way to fill missing values using the nearest neighbors in your dataset.

Instead of using mean, mode, or a fixed value, it looks for similar rows (neighbors) and estimates the missing value based on them.
“If a value is missing, look at the most similar rows and use their values to estimate the missing value.”

“Look at the k most similar data points and use their values to guess the missing one.”

K = number of neighbors to consider

Works for numerical and sometimes categorical data.


**How It Works (Step by Step)**

1. Find the row with a missing value.

2. Compute similarity with all other rows using other columns (like distance).

3. Take the k nearest neighbors (most similar rows).

4. Replace the missing value with:

a. Mean of neighbors (numerical)

b. Mode of neighbors (categorical)

Step 1: Find k=2 nearest neighbors based on Age.

Step 2: Neighbors of Age 19 are Student E (60) and maybe D (20 → close).

Step 3: Take average Marks of neighbors: (60 + 90)/2 = 75

So, Student B Marks = 75

Uses relationships between features

Can be more accurate than mean/mode

Works for both numerical and categorical data

Slower on large datasets

Needs all features to be numerical (or encoded)

Choice of k affects results

```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer

# Sample dataset with missing values
data = {
    "Marks": [80, np.nan, 70, 90, 60],
    "Age": [18, 19, 18, 20, 19]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)
```

Step 1: Create KNN Imputer

```python
# Create KNN Imputer (k=2 nearest neighbors)
imputer = KNNImputer(n_neighbors=2)
print(imputer)
```

Step 2: Fit and Transform Data

```python
# Apply KNN imputation
df_imputed = pd.DataFrame(imputer.fit_transform(df), columns=df.columns)

print("\nAfter KNN Imputation:")
print(df_imputed)
```
