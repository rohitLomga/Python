<h3>1. Take a full name as input and display the initials in uppercase.</h3>

```python
name = input("Enter full name: ")
parts = name.split()
initials = "".join([p[0].upper() for p in parts])
print(initials)
```

<h3>2. Accept a comma-separated list of numbers and calculate their average.</h3>

```python
nums = input("Enter numbers separated by comma: ").split(",")
nums = [float(n) for n in nums]
print("Average:", sum(nums)/len(nums))
```

<h3>3. Take date input in DD-MM-YYYY format and display it as YYYY/MM/DD.</h3>

```python
date = input("Enter date (DD-MM-YYYY): ")
d,m,y = date.split("-")
print(f"{y}/{m}/{d}")
```

<h3>4. Accept a sentence and count the number of words without built-in word count functions.</h3>

```python
sentence = input("Enter sentence: ")
count = 1
for ch in sentence:
    if ch == " ":        
        count += 1
        
print("Word count:", count)
```

<h3>5. Read a number and determine whether it is a palindrome.</h3>

```python
num = input("Enter number: ")
print("Palindrome" if num == num[::-1] else "Not Palindrome")
```

<h3>6. Display a floating-point number rounded to exactly three decimal places.</h3>

```python
num = float(input("Enter number: "))
print(f"{num:.3f}")
```

<h3>7. Align a user-entered string to the center of 30 characters.</h3>

```python
text = input("Enter text: ")
print(f"{text:^30}")
```

<h3>8. Display name, age, and salary in tabular format using string formatting.</h3>

```python
name = "Joe"
age = 30
salary = 25000
print(f"{'Name':10}{'Age':10}{'Salary':10}")
print(f"{name:10}{age:2}{salary:13}")
```

<h3>9. Format and display a number as currency with rupee symbol and two decimals.</h3>

```python
amount = float(input("Enter amount: "))
print(f"₹{amount:.2f}")
```

<h3>10. Display multiple values in one formatted string using positional and named placeholders.</h3>

```python
name = "Rohit"
age = 21
print("Name:{0} Age:{age}".format(name,age=age))
```
