<h3>1 . Write a python function power(a,b) without using ** or pow().</h3>

```python
def power(a, b):
    total = 1

    for x in range(1, b + 1):
        total *= a

    return total

power(2,3)
```

<h3>2 . Write a Python function that takes two tuples representing (x, y) coordinates and calculates the Euclidean distance using math.sqrt.</h3>

```python
import math

point1 = (3, 5)
point2 = (4, 7)

x = (point2[0] - point1[0]) ** 2
y = (point2[1] - point1[1]) ** 2

distance = math.sqrt(x + y)

distance
```

<h3>3 . Write a Python function that takes a list and returns a new list with distinct elements from the first list.</h3>

```python
def unique_list(l):
    new_list = set(l)
    return list(new_list)

list1 = [4, 4, 5, 5, 6, 6]

unique_list(list1)
```

<h3>4 . </h3>

```python
def is_palindrome(text):
    text = text.lower()
    if text == text[::-1]:
        return "Palindrome"
    else:
        return "Not Palindrome"

word = input("Enter a string: ")
print(is_palindrome(word))
```

<h3>7 . </h3>

```python
def is_prime(n):
    if n <= 1:
        return "Not Prime"
    
    for i in range(2, n):
        if n % i == 0:
            return "Not Prime"
    
    return "Prime"

num = int(input("Enter number: "))
print(is_prime(num))
```

<h3>9 .</h3>

```python
def multiply_list(numbers):
    result = 1
    for n in numbers:
        result *= n
    return result

nums = [2, 3, 4, 5]
print("Multiplication:", multiply_list(nums))
```

<h3>10 .</h3>

```python
import math

def find_gcd_lcm(a, b):
    gcd = math.gcd(a, b)
    lcm = (a * b) // gcd
    print("GCD:", gcd)
    print("LCM:", lcm)

find_gcd_lcm(12, 18)
```

<h3>11 .</h3>

```python
import json

json1 = {"name": "Rohit", "age": 21}
json2 = {"age": 22, "city": "Ranchi"}

result = {}

for key in json1:
    result[key] = json1[key]

for key in json2:
    if key in result:
        result[key] = [result[key], json2[key]]
    else:
        result[key] = json2[key]

print(json.dumps(result, indent=4))
```

<h3>14 .</h3>

```python
import re

text = "Contact numbers: 123-456-7890, 987-654-3210"

pattern = r'(\d{3})-(\d{3})-(\d{4})'

masked = re.sub(pattern, r'\1-***-\3', text)

print(masked)
```
