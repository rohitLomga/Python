<h3>3 . Write a program to define a function that returns the square and cube of a number.</h3>

```python
def square_cube(n):
    return {
        "square":n**2,
        "cube":n**3
    }
```

```python
num = square_cube(3)
square = num.get("square")
cube = num["cube"]

print(f"Square {square}, Cube : {cube}")
```

<h3>4 . Write a program to create a function that accepts keyword arguments using *args and calculates their sum.</h3>

```python
def sum(*args):
    total = 0
    
    for x in args:
        total += x

    return total
```

```python
sum(4,2,3,1)
```

<h3>8 . write a program to use the map function to double all elements in a list.</h3>

```python
def double(n):
    return n * 2

numbers = [2,4,5,6,7,9,10]

doubled = map(double,numbers)

print(list(doubled))
```

<h3>9 . Write a program to use the filter function to extract even numbers from a list.</h3>

```python
def even(n):
    return n % 2 == 0

even_no = filter(even, numbers)

list(even_no)
```

<h3>10 . Write a program to convert python object to JSON data.</h3>

```python
import json

data = {
    "name": "Rohit",
    "course": "BTech",
    "city": "Khunti"
}

json_data = json.dumps(data)

print(json_data)
```

<h3>11 . Write a program to convert JSON encoded data into python object.</h3>

```python
json_data = '{"name": "Rohit", "age": 20, "course": "BTech"}'

python_obj = json.loads(json_data)

print(python_obj)
print(type(python_obj))
```

<h3>12 . Write a python program to access only unique key value of a python object.</h3>

```python
data = {
    "a": 10,
    "b": 20,
    "c": 10,
    "d": 30,
    "e": 20
}

unique_values = set(data.values())

print("Unique values:", unique_values)
```
