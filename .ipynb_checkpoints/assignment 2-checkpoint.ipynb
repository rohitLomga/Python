```python
import pandas as pd
import numpy as np
```

```python
vehicles = ["Car","Bus","Bicycle","Bike"]
```

```python
vehicle = ["Car","Bus","Bicycle","Bike","Bicycle","Bike","Bicycle","Bike","Bicycle","Bicycle"]

labels, unique_vehicle = pd.factorize(vehicle)
labels = labels + 1

# print("Labels:", labels)
# print("Unique values:", uniques)

binary_data = [format(i, '03b') for i in labels]

matrix = np.array([list(map(int, list(b))) for b in binary_data])

print(f"Vechicles {unique_vehicle}")
print(f"Binary {matrix}")

```

```python
first_names = ["Rohit","Aman","Priya","Neha","Arjun",
               "Kiran","Sonal","Rahul","Anjali","Vikas",
               "Meena","Suresh","Pooja","Deepak","Riya",
               "Aakash","Nisha","Varun","Sneha","Tarun",
               "Isha","Manoj","Kavya","Ramesh","Divya",
               "Harsh","Komal","Aditya","Simran","Yash"]

last_names = ["Sharma","Verma","Singh","Kapoor","Mehta",
              "Yadav","Gupta","Das","Kumar","Pandey",
              "Rao","Joshi","Patel","Nair","Chopra",
              "Saxena","Malhotra","Bose","Trivedi","Shah",
              "Tiwari","Bhat","Gill","Chaudhary","Reddy",
              "Sinha","Arora","Ghosh","Kulkarni","Mishra"]

states = ["Jharkhand","Bihar","Delhi","Maharashtra","UP"]

data = {
    "First_Name": first_names,
    "Last_Name": last_names,
    "State": np.random.choice(states, 30),
    "Owning_Vehicle": np.random.choice(vehicles, 30)
}

df = pd.DataFrame(data)

print("Original Dataset:\n")
# print(df)
# df

# all_data = pd.

labels, unique_vehicle = pd.factorize(df['Owning_Vehicle'])
labels = labels + 1

# print("Labels:", labels)
# print("Unique values:", uniques)

binary_data = [format(i, '03b') for i in labels]

matrix = np.array([list(map(int, list(b))) for b in binary_data])

print(f"Unique Vehicles {unique_vehicle}")
print(f"All Data {matrix}")
```

```python
labels, uniques = pd.factorize(df["Owning_Vehicle"])

df["Vehicle_Label"] = labels

print("\nEncoded Dataset:\n")
print(df)

print("\nUnique Vehicle Categories:")
print(uniques)
```

```python
df_encoded = pd.get_dummies(df, columns=["Owning_Vehicle"], dtype=int)


pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.expand_frame_repr', False) 


print("\nAfter One-Hot Encoding:\n")
print(df_encoded)
```

```python
df["Owning_Vehicle_Ordinal"] = pd.Categorical(
    df["Owning_Vehicle"],
    categories=vehicles,
    ordered=True
).codes

# print(df.head())
df
```
