###**What is data scaling?**

Data scaling means changing the numbers in your data so they are on a similar scale (range), without changing what they actually mean.

Changing the values of data into a similar range (scale) without changing their meaning.


Example:
If you measure:

a. height in centimeters (e.g., 170)

b. weight in kilograms (e.g., 65)

The numbers are very different in size, even though both are important. Scaling helps make them fair to compare.

###**Why do we need data scaling?**

Many machine learning algorithms:

a. compare numbers,

b. calculate distances,

c. or try to find patterns using math.

If one feature has big numbers and another has small numbers, the big one can unfairly dominate the model.

Scaling fixes that.

###**Here is the data that usually doesn't need scaling:**

**1. Categorical Data (Encoded)**

If you have already transformed categories into numbers (like One-Hot Encoding), you usually don't scale them further.

Why: One-hot encoded columns are already binary (0 or 1). Scaling a 0 or 1 doesn't change anything; they are already on a perfect "mini-scale."

Example: A column Is_Male where 1 = Yes and 0 = No. Scaling this would be redundant.

**2. Binary Features**

Any data that only has two states (True/False, 0/1, Yes/No).

Why: These are already at the "base" level. There is no "distance" to compress because there are only two points.

**3. Tree-Based Algorithms**

 This is the most common scenario. Models like Decision Trees, Random Forests, and XGBoost don't care about the scale of your numbers.

 Why: These models split data based on "greater than" or "less than" logic at certain thresholds.

**4.Ordinal Data with Meaningful Ranks**

When your numbers represent a specific rank where the "gap" between numbers is the same and intentional.

Why: If you have a survey scale of 1 to 5 (Star Ratings), the scale is already constrained and meaningful.

Example: A 5-star review is exactly 5 times "better" than a 1-star review in a very specific, human-defined way. Scaling might turn these into decimals that are harder to interpret.

**5. Cyclical features**
Data with a cyclical nature often benefits more from cyclical encoding rather than standard scaling techniques.

###**Data that Usually needs Scaling:**

**1. Continuous Numerical Features with Wide Ranges**

 It refers to data that can take any value within a specific interval, where the difference between the smallest (minimum) and largest (maximum) values is massive.

 Continuous: The values aren't just whole numbers like 1, 2, 3. They can be $1.5$, $10.75$, or $999.99$.

 Wide Range: The spread (or variance) is so large that the average value doesn't really represent the data well because the extremes are so far apart.

 **2. Features are measured in different units**

 When we say features are measured in different units, we mean that the columns in your spreadsheet are talking about completely different things—like weight, distance, money, or time.

The problem is that computers don't understand what a "unit" is. They just see raw numbers. If you don't fix this, the computer will try to compare apples to airplanes and get very confused.
example: Imagine you are building an AI to predict how much a used car should cost. You give it two pieces of information:

Mileage: 150,000 (measured in miles)

Number of Previous Owners: 2 (measured in people)

The Computer's Mistake: To a human, 150,000 miles is a lot, and 2 owners is a little. But to a computer, 150,000 is mathematically 75,000 times "bigger" than 2. Because the unit of miles naturally results in larger numbers, the AI might assume that mileage is 75,000 times more important than the number of owners, simply because the numbers are bigger.


**3.Significantly different magnitudes**

It is the difference between a giant and a gnat living in the same dataset. Even if both features are important, the giant (the large magnitude) tends to step on the gnat (the small magnitude).

**4. Percentage or Ratio features**

A percentage is a value expressed out of 100.

a. 75% attendance

b. 12% interest rate

A ratio compares two quantities.

a. Debt-to-income ratio

b. Male-to-female ratio

c. Price-to-earnings ratio

**5. Skewed distribution**

A skewed distribution means your data is "unbalanced" or leaning heavily to one side.A skewed distribution matters because scaling methods behave differently depending on how the data is distributed.

###**Main reasons for scaling data :**
**1. To avoid one feature dominating others**
Some features have big numbers, others have small numbers.

Example:

Age: 18 – 60

Income: 20,000 – 1,00,000

Without scaling, the model may think income is more important just because the numbers are bigger.

Scaling puts them on a similar range.

**2.  To improve model accuracy**

Many machine learning algorithms work better when features are scaled.

Scaling helps the model:

learn faster

find better patterns

make more accurate predictions.

**3. To make distance-based algorithms work properly**

Algorithms like:

a. KNN

b. K-means

c. SVM

calculate distance between data points.

Without scaling: large-value features dominate distance

results become incorrect.

**4. To speed up training**

For models that use optimization (like gradient descent):

scaling helps reach the best solution faster

reduces training time

**5. To help algorithms converge properly**

Unscaled data can cause:

unstable training

poor convergence

wrong coefficients

Scaling fixes this.

###**Scaling Methods**

**1. Min-Max Scaling (Normalization)**

This is the most straightforward method. It squishes all your data into a fixed range, usually 0 to 1.

How it works: It takes the smallest value and makes it 0, takes the largest value and makes it 1, and spreads everything else in between.

Best for: When you know the boundaries of your data (like pixels 0-255) or when your model requires values between 0 and 1 (like Image Processing or Neural Networks).

The Flaw: It is very sensitive to outliers. If you have one "giant" number, it will squish all your "normal" numbers into a tiny space near zero.

MinMaxScaler

This is a class used to scale data between 0 and 1 (by default).


Formula used:

Xscaled = (X - Xmin)/Xmax -Xmin

It transforms values so that:

Minimum value → 0

Maximum value → 1

All other values → between 0 and 1

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
from scipy import stats

# Read the data
data = {
    'Temperature_Celsius': [15, 18, 22, 25, 28, 30, 32, 29, 26, 23, 20, 17],
    'Humidity_Percent': [50, 55, 60, 65, 70, 75, 80, 72, 68, 62, 58, 52],
    'Wind_Speed_kmh': [5, 8, 12, 15, 10, 7, 20, 18, 14, 9, 6, 11],
    'Golfers_Count': [20, 35, 50, 75, 100, 120, 90, 110, 85, 60, 40, 25],
    'Green_Speed': [8.5, 9.0, 9.5, 10.0, 10.5, 11.0, 11.5, 11.0, 10.5, 10.0, 9.5, 9.0]
}

df = pd.DataFrame(data)
```

```python
#Min-Max Scaling for Temperature_Celsius
min_max_scaler = MinMaxScaler()
df['Temperature_MinMax'] = min_max_scaler.fit_transform(df[['Temperature_Celsius']])
print(df)
```

MinMaxScaler() → Creates a scaler object that can scale numbers to a range between 0 and 1.

min_max_scaler → The variable name where we store this scaler.

df[['Temperature_Celsius']]

Selects the column Temperature_Celsius as a DataFrame (double brackets [[ ]] keep it 2D).

min_max_scaler.fit_transform(...)

fit() → Finds the minimum and maximum values of Temperature_Celsius.

transform() → Applies the formula:

Converts the original temperature values into 0–1 range.

df['Temperature_MinMax'] = ...

Stores the scaled values in a new column called Temperature_MinMax.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# 1. Create a sample dataset
data = {
    'Age': [25, 30, 45, 50, 22, 35, 60],
    'Salary': [45000, 50000, 80000, 110000, 38000, 60000, 150000]
}
df = pd.DataFrame(data)

# 2. Initialize the MinMaxScaler (defaults to 0 to 1 range)
scaler = MinMaxScaler()

# 3. Fit and Transform the data
# This calculates the min/max and applies the formula to every row
df_scaled = pd.DataFrame(scaler.fit_transform(df), columns=df.columns)

print("Original Data:")
print(df)
print("\nScaled Data (0 to 1 range):")
print(df_scaled.round(2))
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt

# 1. Create a sample dataset
data = {
    'Age': [25, 30, 45, 50, 22, 35, 60],
    'Salary': [45000, 50000, 80000, 110000, 38000, 60000, 150000]
}
df = pd.DataFrame(data)

# 2. Initialize the MinMaxScaler
scaler = MinMaxScaler()

# 3. Fit and transform the data
# We select the columns to scale
df_scaled = df.copy()
df_scaled[['Age', 'Salary']] = scaler.fit_transform(df[['Age', 'Salary']])

# Round for easier reading in the response
df_scaled_rounded = df_scaled.round(2)

# Save to CSV as per guidelines
df_scaled.to_csv('/content/min_max_scaled_data.csv', index=False)

# 4. Visualization
plt.figure(figsize=(12, 5))

# Plot Before Scaling
plt.subplot(1, 2, 1)
plt.scatter(df['Age'], df['Salary'], color='blue')
plt.title('Before Scaling')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.grid(True)

# Plot After Scaling
plt.subplot(1, 2, 2)
plt.scatter(df_scaled['Age'], df_scaled['Salary'], color='green')
plt.title('After Min-Max Scaling (0 to 1)')
plt.xlabel('Age (Scaled)')
plt.ylabel('Salary (Scaled)')
plt.grid(True)

plt.tight_layout()
plt.savefig('min_max_scaling_comparison.png')

# Output the dataframes for the response
print("Original Data:")
print(df)
print("\nScaled Data (Min-Max):")
print(df_scaled_rounded)
```

**2. Standard Scaling** (also called Standardization or Z-score Normalization) : It is a method of resizing data so that it centers around the average.Standard Scaling centers data around a mean of 0 and scales it to a standard deviation of 1, achieved by subtracting the mean and dividing by the standard deviation.

Standard deviation measures how spread out the numbers in a dataset are.

a. If the numbers are close to the mean, the standard deviation is small.

b. If the numbers are far from the mean, the standard deviation is large.

Unlike Min-Max scaling, which squishes everything into a 0-to-1 box, Standardization looks at how many "steps" (standard deviations) a data point is away from the mean.

Imagine you are comparing the heights of students in two different countries. Instead of using centimeters, you describe each student by saying:

"This student is exactly average" (Score = 0).

"This student is above average" (Positive Score).

"This student is below average" (Negative Score).

Step 1: Formula for Standardization

Standardization formula:

Z =( X-μ )/σ
	​


Where:


X = original value


μ = mean


σ = standard deviation


Z = standardized value



```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
from scipy import stats

# Read the data
data = {
    'Temperature_Celsius': [15, 18, 22, 25, 28, 30, 32, 29, 26, 23, 20, 17],
    'Humidity_Percent': [50, 55, 60, 65, 70, 75, 80, 72, 68, 62, 58, 52],
    'Wind_Speed_kmh': [5, 8, 12, 15, 10, 7, 20, 18, 14, 9, 6, 11],
    'Golfers_Count': [20, 35, 50, 75, 100, 120, 90, 110, 85, 60, 40, 25],
    'Green_Speed': [8.5, 9.0, 9.5, 10.0, 10.5, 11.0, 11.5, 11.0, 10.5, 10.0, 9.5, 9.0]
}

df = pd.DataFrame(data)
```

```python
std_scaler = StandardScaler()
df['Wind_Speed_Standardized'] = std_scaler.fit_transform(df[['Wind_Speed_kmh']])
print(df.round(0))
```

**3. Robust scaling**

Robust Scaling centers data around the median and scales using the interquartile range (IQR)

It is a way to scale data that is not affected by outliers (very large or very small values).

It uses:

a. Median

b. Interquartile Range (IQR)
instead of mean and standard deviation.

**Goals:**
- Handle outliers effectively without being overly influenced by them.
- Maintain the relative order of data points.
- Achieve a stable scaling in the presence of noisy data.


Example outlier:

Normal salaries: 30,000 – 60,000

One salary: 1,000,000

Normal scaling methods get distorted by this extreme value.

Robust scaling handles this problem.


Apply Robust Scaling Formula:

Xscaled = (X - Median)/IQR

####**How robust scaling works**

**(value − median) / IQR**

Where:

Median = middle value

IQR = Q3 − Q1 (middle 50% of data)




Original data (Salary in $)
30,000    35,000   40,000   45,000   1,000,000


Here:

1,000,000 is an outlier


**Step 1**: Find median:Median = 40,000

**Step 2**: Find IQR

Q1 = 35,000

Q3 = 45,000

IQR = 45,000 − 35,000 = 10,000

**Step 3**: Apply robust scaling

For 30,000:

(30,000 − 40,000) / 10,000 = -1


For 45,000:

(45,000 − 40,000) / 10,000 = 0.5


For 1,000,000:

(1,000,000 − 40,000) / 10,000 = 96

```python
robust_scaler = RobustScaler()
df['Humidity_Robust'] = robust_scaler.fit_transform(df[['Humidity_Percent']])
print(df.round(2))
```

```python
import pandas as pd
from sklearn.preprocessing import RobustScaler
# Example humidity data with an outlier
df = pd.DataFrame({
    'Humidity_Percent': [30, 35, 40, 45, 100]
})

#print("Original Data:")
#print(df)
robust_scaler = RobustScaler()#Create the RobustScaler Object
df['Humidity_Robust'] = robust_scaler.fit_transform(df[['Humidity_Percent']])#Fit and Transform the Data
print("\nScaled Data (RobustScaler):")
print(df)


```

**4. Log Transformation**

Log transformation means applying a logarithm (usually log base 10 or natural log) to your data.
It shrinks large values and spreads small values.

We use log transformation to:

1 Handle skewed data

2 Reduce the effect of outliers

3 Make data more normal (bell-shaped)

4 Improve performance of some ML models

```python
import pandas as pd
import numpy as np

# Sample income data
df = pd.DataFrame({'Income': [1000, 2000, 3000, 4000, 100000]})

# Apply log transformation
df['Income_log'] = np.log(df['Income'])

print(df)
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
from scipy import stats

# Read the data
data = {
    'Temperature_Celsius': [15, 18, 22, 25, 28, 30, 32, 29, 26, 23, 20, 17],
    'Humidity_Percent': [50, 55, 60, 65, 70, 75, 80, 72, 68, 62, 58, 52],
    'Wind_Speed_kmh': [5, 8, 12, 15, 10, 7, 20, 18, 14, 9, 6, 11],
    'Golfers_Count': [20, 35, 50, 75, 100, 120, 90, 110, 85, 60, 40, 25],
    'Green_Speed': [8.5, 9.0, 9.5, 10.0, 10.5, 11.0, 11.5, 11.0, 10.5, 10.0, 9.5, 9.0]
}

df = pd.DataFrame(data)
df['Golfers_Log'] = np.log1p(df['Golfers_Count'])
print(df)
```

```python
df['Golfers_Log'] = np.log10(df['Golfers_Count'])
print(df)
```

Natural log: ln(X) → np.log(X) in Python

Base 10 log: log10(X) → np.log10(X) in Python

Log1p (log(1+X)) → Useful if data contains 0

```python
import pandas as pd
import numpy as np

# Sample data
df = pd.DataFrame({'Income': [1000, 2000, 3000, 4000, 100000]})

# Natural log transformation
df['Income_log'] = np.log(df['Income'])

# Base 10 log transformation (optional)
df['Income_log10'] = np.log10(df['Income'])

# Log1p transformation if data has 0s
df['Income_log1p'] = np.log1p(df['Income'])

print(df)
```

**5. Box-Cox Transformation**

This is a family of power transformations (that includes log transformation as a special case) that aims to normalize the distribution of data by applying a power transformation with a parameter lambda (λ), which is optimized to achieve the desired normality.
Features needing normalization to approximate a normal distribution.

Box-Cox transformation is a method used to:

a. make data more normal (bell-shaped)

b. reduce skewness

c. improve model performance

Example:

**Original data** : 10, 20, 30, 50, 200

Most values are small

One large value (200)

Data is skewed to the right.

**After Box-Cox transformation**: 1.5, 2.0, 2.3, 2.8, 3.2

Values are closer together

Distribution becomes more balanced

Outlier impact is reduced

**How Box-Cox works**

Box-Cox uses a parameter called λ (lambda)

λ = 1 → no change

λ = 0 → log transformation

λ = 0.5 → square root

λ = −1 → inverse

The algorithm finds the best λ automatically.



**Key Points About λ**

λ = 1 → No transformation (original data stays the same)

λ = 0 → Becomes log transformation

0 < λ < 1 → Compresses large values (reduces skewness)

λ > 1 → Expands large values

λ < 0 → Inverts data in a certain way

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
from scipy import stats

# Read the data
data = {
    'Temperature_Celsius': [15, 18, 22, 25, 28, 30, 32, 29, 26, 23, 20, 17],
    'Humidity_Percent': [50, 55, 60, 65, 70, 75, 80, 72, 68, 62, 58, 52],
    'Wind_Speed_kmh': [5, 8, 12, 15, 10, 7, 20, 18, 14, 9, 6, 11],
    'Golfers_Count': [20, 35, 50, 75, 100, 120, 90, 110, 85, 60, 40, 25],
    'Green_Speed': [8.5, 9.0, 9.5, 10.0, 10.5, 11.0, 11.5, 11.0, 10.5, 10.0, 9.5, 9.0]
}

df = pd.DataFrame(data)

df['Green_Speed_BoxCox'], lambda_param = stats.boxcox(df['Green_Speed'])
print(df)
```

stats.boxcox(...)

stats.boxcox is a function from the scipy library that applies the Box-Cox transformation.

What it does internally:

Finds the best lambda (
𝜆
λ) that makes your data as normally distributed as possible.

Applies the Box-Cox formula:

lambda_param = ...

The Box-Cox function also returns the best lambda value.

This lambda_param tells you which exponent λ was used to transform the data.

For example:

If lambda_param = 0 → It applied a log transformation.

If lambda_param = 0.3 → Slightly compresses large values to reduce skewness.

```python
from scipy import stats
import pandas as pd

df = pd.DataFrame({'Green_Speed': [10, 15, 20, 25, 100]})

df['Green_Speed_BoxCox'], lambda_param = stats.boxcox(df['Green_Speed'])

print(df.round(2))

```

```python
import numpy as np
from scipy.stats import boxcox
data = np.array([10, 20, 30, 50, 200])

print("Original data:")
print(data)
transformed_data, best_lambda = boxcox(data)

print("\nBox-Cox transformed data:")
print(transformed_data.round(2))

print("\nBest lambda value:")
print(best_lambda.round(2))
```

###**Conclusion**

```python
import numpy as np
import pandas as pd

# Sample dataset
data = pd.DataFrame({
    "Salary": [30000, 35000, 40000, 45000, 1000000]
})

print("Original Data:")
print(data)
from sklearn.preprocessing import MinMaxScaler

minmax_scaler = MinMaxScaler()
data["MinMax_Scaled"] = minmax_scaler.fit_transform(data[["Salary"]])

print("\nMin-Max Scaled Data:")
print(data[["Salary", "MinMax_Scaled"]])
from sklearn.preprocessing import StandardScaler

standard_scaler = StandardScaler()
data["Standard_Scaled"] = standard_scaler.fit_transform(data[["Salary"]])

print("\nStandard Scaled Data:")
print(data[["Salary", "Standard_Scaled"]])
from sklearn.preprocessing import RobustScaler

robust_scaler = RobustScaler()
data["Robust_Scaled"] = robust_scaler.fit_transform(data[["Salary"]])

print("\nRobust Scaled Data:")
print(data[["Salary", "Robust_Scaled"]])
data["Log_Transformed"] = np.log(data["Salary"] + 1)

print("\nLog Transformed Data:")
print(data[["Salary", "Log_Transformed"]])
from scipy.stats import boxcox

data["BoxCox_Transformed"], lambda_value = boxcox(data["Salary"])

print("\nBox-Cox Transformed Data:")
print(data[["Salary", "BoxCox_Transformed"]])

print("\nBest Lambda Value:", lambda_value)
print("\nFinal Dataset with All Transformations:")
print(data)
```
