<h3>1. Create Student class with name and age using __init__.</h3>

```python
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age

s = Student("Rohit",21)
print(s.name,s.age)
```

<h3>2. Create Car class with brand and model.</h3>

```python
class Car:
    def __init__(self,brand,model):
        self.brand = brand
        self.model = model
        
c = Car("Toyota","Camry")
print(c.brand,c.model)
```

<h3>4. Employee class with name id salary.</h3>

```python
class Employee:
    def __init__(self,name,id,salary):
        self.name = name
        self.id = id
        self.salary = salary
        
e = Employee("Rohit",1,25000)
print(e.name,e.salary)
```

<h3>7. Student class with private marks and getter.</h3>

```python
class Student:
    def __init__(self,marks):
        self.__marks = marks
    def get_marks(self):
        return self.__marks
            
s = Student(90)
print(s.get_marks())
```

<h3>8. BankAccount with private balance setter.</h3>

```python
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance
    def set_balance(self,b):
        self.__balance = b
    def get_balance(self):
        return self.__balance 
        
a = BankAccount(1000)
a.set_balance(2000)
print(a.get_balance())
```

<h3>10. Employee class with setter salary.</h3>

```python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    
    def set_salary(self,s):
        self.salary = s
        
e = Employee("Rohit",20000)
e.set_salary(25000)
print(e.salary)
```

<h3>11. Calculator class with add subtract multiply divide.</h3>

```python
class Calculator:
    def add(self,a,b):
        return a + b
        
    def sub(self,a,b):
        return a-b
        
    def mul(self,a,b):
        return a * b
        
    def div(self,a,b):
        return a/b
        
c = Calculator()
print(c.add(5,3))
```

<h3>12. Student class with display method.</h3>

```python
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    
    def display(self):
        print(self.name,self.age)
        
s = Student("Rohit",21)
s.display()
```

<h3>13. Rectangle class with area method.</h3>

```python
class Rectangle:
    def __init__(self,l,b):
        self.l = l
        self.b = b
    
    def area(self):
        return self.l * self.b

r=Rectangle(10,5)
print(r.area())
```

<h3>14. Circle class with circumference.</h3>

```python
import math
class Circle:
    def __init__(self,r):
        self.r = r
        
    def circumference(self):
        return 2 * math.pi * self.r
        
c = Circle(5)
print(c.circumference())
```
