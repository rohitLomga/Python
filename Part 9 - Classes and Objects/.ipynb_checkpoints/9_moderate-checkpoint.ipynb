<h3>1. Employee class that displays details.</h3>

```python
class Employee:
    def __init__(self,name,id,salary):
        self.name = name
        self.id = id
        self.salary = salary
        
    def display(self):
        print(self.name,self.id,self.salary)
        
e = Employee("Rohit",1,30000)
e.display()
```

<h3>2. Book class prints formatted details.</h3>

```python
class Book:
    def __init__(self,title,author,price):
        self.title = title
        self.author = author
        self.price = price
        
    def show(self):
        print(f"{self.title} by {self.author} costs {self.price}")
        
b = Book("Python","Guido",500)
b.show()
```

<h3>3. Rectangle class area and perimeter.</h3>

```python
class Rectangle:
    def __init__(self,l,b):
        self.l = l
        self.b = b
        
    def area(self):
        return self.l*self.b
        
    def perimeter(self):
        return 2*(self.l+self.b)
        
r = Rectangle(10,5)
print(r.area(),r.perimeter())
```

<h3>4. Car class formatted details.</h3>

```python
class Car:
    def __init__(self,brand,model,year):
        self.brand = brand
        self.model = model
        self.year = year
        
    def display(self):
        print(f"{self.year} {self.brand} {self.model}")
        
        
c = Car("Toyota","Camry",2022)
c.display()
```

<h3>5. BankAccount with private balance getter setter.</h3>

```python
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance
        
    def get_balance(self):
        return self.__balance
        
    def set_balance(self,b):
        if b >= 0:
            self.__balance = b
            
a = BankAccount(1000)
a.set_balance(2000)
print(a.get_balance())
```
