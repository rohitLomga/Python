**3.Write a Python program to create a class Rectangle that initializes length and breadth and calculates both area and perimeter using methods.**

```python
class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth

    def perimeter(self):
        return 2 * (self.length + self.breadth)


# create object
r1 = Rectangle(10, 5)

print("Length:", r1.length)
print("Breadth:", r1.breadth)
print("Area:", r1.area())
print("Perimeter:", r1.perimeter())
```

**5.Write a Python program to create a class BankAccount with a private attribute balance. Provide getter and setter methods to read and update the balance safely.**

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # private attribute

    # getter method
    def get_balance(self):
        return self.__balance

    # setter method
    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
            print("Balance updated successfully.")
        else:
            print("Invalid balance amount.")

# create object
account = BankAccount(5000)

# read balance
print("Current Balance:", account.get_balance())

# update balance
account.set_balance(8000)

# read updated balance
print("Updated Balance:", account.get_balance())
```

**7.Write a Python program to create a class Temperature with a property celsius. Add another property fahrenheit to convert Celsius to Fahrenheit automatically.**

```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    @property
    def fahrenheit(self):
        return (self.celsius * 9/5) + 32


# create object
t1 = Temperature(25)

print("Temperature in Celsius:", t1.celsius)
print("Temperature in Fahrenheit:", t1.fahrenheit)
```

**9.Write a Python program to create a class Circle with a private attribute radius and a method to calculate area and circumference, using properties to access radius.**

```python
import math

class Circle:
    def __init__(self, radius):
        self.__radius = radius   # private attribute

    # getter property
    @property
    def radius(self):
        return self.__radius

    # setter property
    @radius.setter
    def radius(self, value):
        if value > 0:
            self.__radius = value
        else:
            print("Radius must be positive")

    def area(self):
        return math.pi * self.__radius ** 2

    def circumference(self):
        return 2 * math.pi * self.__radius


# create object
c1 = Circle(7)

print("Radius:", c1.radius)
print("Area:", c1.area())
print("Circumference:", c1.circumference())
```

**11.Write a Python program to create a class Student with a method display_grade() to assign grades based on marks in multiple subjects.**

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks   # list of marks

    def display_grade(self):
        total = sum(self.marks)
        avg = total / len(self.marks)

        if avg >= 90:
            grade = "A"
        elif avg >= 75:
            grade = "B"
        elif avg >= 60:
            grade = "C"
        elif avg >= 50:
            grade = "D"
        else:
            grade = "F"

        print("Student Name:", self.name)
        print("Marks:", self.marks)
        print("Average:", avg)
        print("Grade:", grade)


# create object
marks = [85, 78, 92, 74]
s1 = Student("Rahul", marks)

# display grade
s1.display_grade()
```

**12.Write a Python program to create a class Library with methods to:**
*add books*

*remove books*

*search books by title*

```python
class Library:
    def __init__(self):
        self.books = []

    def add_book(self, title):
        self.books.append(title)
        print(f'"{title}" added to library.')

    def remove_book(self, title):
        if title in self.books:
            self.books.remove(title)
            print(f'"{title}" removed from library.')
        else:
            print("Book not found.")

    def search_book(self, title):
        if title in self.books:
            print(f'"{title}" is available in the library.')
        else:
            print(f'"{title}" is not available in the library.')


# create library object
lib = Library()

# operations
lib.add_book("Python Basics")
lib.add_book("Data Structures")

lib.search_book("Python Basics")

lib.remove_book("Data Structures")

lib.search_book("Data Structures")
```

**13.Write a Python program to create a class Account with methods:**

*deposit()*

*withdraw()*

*check_balance()*

```python
class Account:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient Balance")

    def check_balance(self):
        print("Current Balance:", self.balance)


# create object
acc = Account(1000)

acc.deposit(500)
acc.withdraw(300)
acc.check_balance()
```

**14.Write a Python program to create a class Employee with a method increment_salary() that increases salary based on performance rating.
Include input validation and formatted output.**

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def increment_salary(self, rating):
        if rating == 5:
            self.salary += self.salary * 0.20
        elif rating == 4:
            self.salary += self.salary * 0.15
        elif rating == 3:
            self.salary += self.salary * 0.10
        elif rating == 2:
            self.salary += self.salary * 0.05
        elif rating == 1:
            self.salary += self.salary * 0.02
        else:
            print("Invalid rating. Rating must be between 1 and 5.")
            return
        
        print(f"Updated salary for {self.name}: ₹{self.salary:.2f}")


# Input
name = input("Enter employee name: ")

try:
    salary = float(input("Enter current salary: "))
    rating = int(input("Enter performance rating (1-5): "))
    
    if salary < 0:
        print("Salary cannot be negative.")
    else:
        emp = Employee(name, salary)
        emp.increment_salary(rating)

except ValueError:
    print("Invalid input. Please enter numeric values for salary and rating.")
```
