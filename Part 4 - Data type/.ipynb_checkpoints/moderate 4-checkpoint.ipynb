<h3>1 . Write a python function that takes a list of words and return the longest word abd the length of the longest one.</h3>

```python
def longest_word(words):
    longest = max(words, key=len)
    return longest, len(longest)

words = ["apple", "banana", "grapefruit", "kiwi"]
print(longest_word(words))
```

<h3>2 . Write a python program to change a given string to a newly string where the first and last chars have been exchanged.</h3>

```python
def swap_first_last(s):
    if len(s) < 2:
        return s
    return s[-1] + s[1:-1] + s[0]

print(swap_first_last("Python"))
```

<h3>3 . Write a python program to remove duplicates from a list.</h3>

```python
def remove_duplicates(lst):
    return list(set(lst))

numbers = [1, 2, 2, 3, 4, 4, 5]
print(remove_duplicates(numbers))
```

<h3>4 . Write a python function that takes two lists and returns True if they have atleast one common member.</h3>

```python
def has_common(list1, list2):
    return bool(set(list1) & set(list2))

print(has_common([1, 2, 3], [3, 4, 5]))
```

<h3>5 . Write a Python program to calculate the difference between the two lists.</h3>

```python
def list_difference(list1, list2):
    return list(set(list1) - set(list2))

print(list_difference([1, 2, 3, 4], [2, 4]))
```

<h3>6 . Write a Python program to find the second largest number in a list.</h3>

```python
def second_largest(lst):
    unique = list(set(lst))
    unique.sort()
    return unique[-2]

print(second_largest([10, 20, 4, 45, 99]))
```

<h3>7 . Write a Python program to convert a list of multiple integers into a single integer.</h3>

```python
def list_to_integer(lst):
    return int("".join(map(str, lst)))

print(list_to_integer([1, 2, 3, 4]))
```

<h3>8 . Write a Python function to find the maximum sum sub-sequence in a list. Return the maximum value.</h3>

```python
def max_subsequence_sum(lst):
    max_sum = current_sum = lst[0]
    
    for num in lst[1:]:
        current_sum = max(num, current_sum + num)
        max_sum = max(max_sum, current_sum)
    
    return max_sum

print(max_subsequence_sum([-2,1,-3,4,-1,2,1,-5,4]))
```

<h3>9 . Write a python script to generate and print a dictionary that contains a number (between 1 and n) in the form (x,x*x).</h3>

```python
n = 5
result = {x: x*x for x in range(1, n+1)}
print(result)
```

<h3>10 . Write a Python program to get the maximum and minimum values of a dictionary.</h3>

```python
def max_min_dict(d):
    return max(d.values()), min(d.values())

data = {'a': 10, 'b': 25, 'c': 5}
print(max_min_dict(data))
```

<h3>11 . A Python Dictionary contains List as a value. Write a Python program to update the list values in the said dictionary.</h3>

```python
data = {
    "numbers": [1, 2, 3],
    "letters": ["a", "b"]
}

data["numbers"].append(4)
data["letters"].append("c")

print(data)
```

<h3>12 . Write a Python program to get the total length of all values in a given dictionary with string values.</h3>

```python
def total_string_length(d):
    return sum(len(value) for value in d.values() if isinstance(value, str))

data = {"a": "hello", "b": "world", "c": "python"}
print(total_string_length(data))
```
