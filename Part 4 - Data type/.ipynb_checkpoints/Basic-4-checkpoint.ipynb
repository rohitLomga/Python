<h3>Write a program to create a list, add elements, and display its contents</h3>

```python
fruits = [
    "Apple",
    "Orange",
    "Cherry",
    "Banana",
    "Pineapple",
    "Grapes",
    "Watermelon",
    "Papaya",
    "Strawberry"
]

print("Before inserting", fruits)

fruits.append("Mango")
print("After Inserting",fruits)
```

<h3>3 . Write a program to slice to extract the first three and last three elements</h3>

```python
first_three = fruits[:3]
last_three = fruits[-3:]

all_six = [first_three, last_three]

all_six
```

<h3>4 . Write a program to unpack a list into individual variables and display them </h3>

```python
fruit1 = fruits[0]
fruit2 = fruits[1]
fruit3 = fruits[3]

print(f"Fruit 1 : {fruit1}, \n Fruit2 : {fruit2} \n Fruit3 : {fruit3}")
```

<h3>6 . Write a program to add and remove items from a list, then display the updated list</h3>

```python
vehicles = ["Car", "Bus", "Bike"]
print("Original List", vehicles)

vehicles.append("Truck")
print("After adding", vehicles)

vehicles.remove("Bike")
print("After removing", vehicles)
```

<h3>7 . Write a program to check if a specific item exists in a list and display its index if found</h3>

```python
search_fruit = "Papaya"

if search_fruit in fruits:
    index = fruits.index(search_fruit)
    print(index)
else: print("Not Found")
```

<h3>10 . Write a program to create a list of squares of numbers from 1 to 10 using list comprehension.</h3>

```python
squares = [x**2 for x in range(1, 11)]

print(squares)
```

<h3>11 . Write a program to combine two lists using the zip function and display the paired elements</h3>

```python
combined = list(zip(fruits, squares))
print(combined)
```

<h3>12 . Write a program to create a set, add elements, and perform operations like union and intersection.</h3>

```python
new_set = {4, 5, 8, 6, 15, 19}
second_set = {100, 200, 300}

print("Original",new_set)

new_set.add(50)
print("Added",new_set)

union_set = new_set.union(second_set)
# union_set = new_set | second_set

print("Union Set",union_set)

intersection_set = new_set.intersection(second_set)
# intersection_set = new_set & second_set

print("Intersection Set",intersection_set)
```
