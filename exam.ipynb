```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

data = [12, 18, 25, 30, 35]

df = pd.DataFrame(data)

std_scaler = StandardScaler()

scaled_data = std_scaler.fit_transform(df)

scaled_data
```

<h3>2</h3>

```python
person = {
    "name": ["Rohit","Raj","Rajesh","Rakesh","Rani","Nikita"],
    "gender": ["Male", np.nan, "Male", np.nan, "Female", np.nan]
}

df = pd.DataFrame(person)

df["gender"].fillna(df["gender"].mode()[0], inplace=True)

df
```

<h3>4</h3>

```python
data = {
    "ID":[1,2,3,4,5],
    "Age":[25, np.nan, 28, 30, np.nan],
    "Salary":[30000, 35000, np.nan, 40000, 38000]
}

df = pd.DataFrame(data)

# 1 . 
print(df.isnull())

print(df.isnull().sum())

# 2
df["Age"].fillna(df["Age"].mean(), inplace=True)

# 3
df["Salary"].fillna(df["Salary"].median(), inplace=True)

df
```
