<h2>Min Max Scaling</h2>

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
from scipy import stats

data = {
    'Rainfall_mm': [2, 5, 12, 20, 35, 50, 65, 40, 25, 15, 8, 3],
    'Solar_Radiation_Wm2': [120, 150, 200, 300, 450, 500, 550, 480, 350, 250, 180, 130],
    'Soil_Moisture_Level': [30, 35, 40, 55, 65, 75, 85, 70, 60, 50, 42, 33],
    'Visitors_Per_Day': [150, 200, 350, 500, 750, 900, 1000, 850, 600, 450, 300, 180],
    'Air_Quality_Index': [45, 50, 55, 60, 70, 80, 95, 85, 75, 65, 58, 48]
}

df = pd.DataFrame(data)
```

```python
min_max_scaler = MinMaxScaler()
df['Visitors_Per_Day_MinMax'] = min_max_scaler.fit_transform(df[['Visitors_Per_Day']])
df
```

<h2>Standard Deviation</h2>

```python
std_scaler = StandardScaler()
df['Visitors_Per_Day_Std'] = std_scaler.fit_transform(df[['Visitors_Per_Day']])
df.round(4)
```

<h2>Robust Scaling</h2>

```python
robust = RobustScaler()
df['robust_rain'] = robust.fit_transform(df[['Rainfall_mm']])
df.round(4)
```
